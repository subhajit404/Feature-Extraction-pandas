```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r'anime.csv')
```

```python
print(df)
```

```python
def Extract_episods(txt):
    check =False
    data = ""
    for i in txt:
        if i==")":
            return data
        
        if check ==True:
            data += i
        if i =='(':
            check=True
    
```

```python
df["Episord"] = df['Title'].apply(Extract_episods)
df["Episord"]=df['Episord'].str.replace(" eps","").astype(int)
```

```python
def time_stamp(txt):
    data=""
    for i in range(len(txt)):
        if txt[i] ==')':
            for j in range(i+1,i+20):
                data +=txt[j]
            return data
```

```python
df['Total_time'] =df['Title'].apply(time_stamp)
df['Total_time']
```
