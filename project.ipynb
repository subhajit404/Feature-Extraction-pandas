```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r'anime.csv')
```

```python
print(df)
```

```python
def Extract_episods(txt):
    check =False
    data = ""
    for i in txt:
        if i==")":
            return data
        
        if check ==True:
            data += i
        if i =='(':
            check=True
    
```

```python
df["Episord"] = df['Title'].apply(Extract_episods)
df["Episord"]=df['Episord'].str.replace(" eps","").astype(int)
```
